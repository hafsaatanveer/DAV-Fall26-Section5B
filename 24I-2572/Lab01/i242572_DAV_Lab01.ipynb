```python
#task1
import pandas as pd
import numpy as np

df = pd.read_csv("adult.csv")

df.head()
df.info()

# missing values
print("Missing values before cleaning:")
print(df.isnull().sum())

df.replace("?", np.nan, inplace=True)

print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

```python
#task2
missing_sum = df.isnull().sum()
missing_percentage = df.isnull().mean() * 100

summary = pd.DataFrame({
    "Missing Count": missing_sum,
    "Missing Percentage": missing_percentage
})
summary = summary.sort_values(
    by="Missing Percentage",
    ascending=False
)
summary
```

```python
#task3

df["workclass"] = df["workclass"].fillna("Unknown")

df["occupation"] = df["occupation"].fillna("Unknown")

df["native.country"] = df["native.country"].fillna("Unknown")

df.isnull().sum()
```

```python
task3 reasoning:
For workclass and occupation, I decided to replace the missing values with "Unknown" instead of deleting the rows. Since these are categorical columns, using "Unknown" keeps the data and does not assume any specific category.

I also replaced the missing values in native.country with "Unknown" for the same reason. There were no missing values in any numerical columns, so no numerical imputation was needed.
```

```python
#TASK4
print("Exact duplicate rows:", df.duplicated().sum())

features = df.drop(columns=['income'])

print("Duplicates ignoring target:", features.duplicated().sum())

print("Decision: Rows with identical attributes but different income outcomes are not treated as duplicates.")
```

```python
#task5


for col in ['education', 'marital.status', 'native.country']:
    print("\nUnique values in", col, "before cleaning:")
    print(df[col].unique())

for col in ['education', 'marital.status', 'native.country']:
    df[col] = df[col].str.strip()

for col in ['education', 'marital.status', 'native.country']:
    print("\nUnique values in", col, "after cleaning:")
    print(df[col].unique())
```

```python
#task6

print("Numerical Descriptive Statistics:")
print(df.describe())

print("\nCategorical Descriptive Statistics:")
print(df.describe(include='object'))

print("\nMean and Median:")
print("Age Mean:", df['age'].mean())
print("Age Median:", df['age'].median())
print("Hours-per-week Mean:", df['hours.per.week'].mean())
print("Hours-per-week Median:", df['hours.per.week'].median())

print("\nComment:")
print("The mean and median are compared to determine whether the distributions are noticeably skewed.")
```

```python
#task 7

print("Most Common Occupation:")
print(df['occupation'].value_counts().head(1))

print("\nSex Percentage Distribution:")
print((df['sex'].value_counts(normalize=True) * 100).round(2))

print("\nIncome Percentage Distribution:")
print((df['income'].value_counts(normalize=True) * 100).round(2))

print("\nIncome Balance:")
income_percent = df['income'].value_counts(normalize=True) * 100

if abs(income_percent.iloc[0] - income_percent.iloc[1]) < 10:
    print("The income classes are relatively balanced.")
else:
    print("The income classes are imbalanced.")
```

```python
#task8

education_check = df.groupby('education')['education.num'].unique()

print("Education and education.num mapping:")
print(education_check)

inconsistencies = education_check[education_check.apply(len) > 1]

print("\nInconsistencies found:")
if len(inconsistencies) == 0:
    print("No inconsistencies found.")
else:
    print(inconsistencies)
```

# Task 10
## 1. Dataset Overview

The dataset used is the UCI Adult Income Dataset. It contains information about individuals from the 1994 US Census and is used to predict whether a person's annual income is greater than $50K.

The original dataset contains **32,561 rows and 15 columns**. It includes **6 numerical columns and 9 categorical columns**.

## 2. Data Quality Issues Found

The dataset contained missing values represented by the `?` symbol:

- `occupation`: 1,843 missing values
- `workclass`: 1,836 missing values
- `native.country`: 583 missing values

There were **24 exact duplicate rows** in the original dataset.

There were also leading/trailing spaces in categorical values, which could create inconsistent categories.

The `education` and `education.num` columns were checked and found to be consistent, with each education level mapping to the same education number.

## 3. Key Observations

The most common occupation was **Prof-specialty**, with **4,136** records.

The sex distribution was:

- Male: **66.92%**
- Female: **33.08%**

The income distribution was:

- `<=50K`: **75.91%**
- `>50K`: **24.09%**

Therefore, the income classes are **imbalanced**, because most individuals belong to the `<=50K` class.

The mean hours-per-week was **40.44 hours**, while the median was **40 hours**. These values are very close, suggesting that the typical working week is around 40 hours.

## 4. Cleaning Decisions Made

The `?` values were treated as missing values and replaced with `NaN`. The missing categorical values were then filled with `Unknown` instead of removing those rows, so that useful observations were not lost.

The **24 exact duplicate rows** were removed.

Leading and trailing whitespace was removed from categorical columns using `.str.strip()`.

No columns were removed. The `education` and `education.num` relationship was checked and found to be consistent.
